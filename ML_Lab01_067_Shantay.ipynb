```python
# Activity B
import numpy as np

marks = np.array([55, 72, 48, 90, 66, 81])

print('Marks:', marks)
print('Mean:', marks.mean())
print('Highest:', marks.max())
print('Passed:', marks[marks >= 50])

print('Minimum:', marks.min())
print('Standard Deviation:', marks.std())

bonus_marks = marks + 5
print('Bonus Marks:', bonus_marks)
print('Original Marks:', marks)

passed_count = np.sum(marks >= 50)
print('Passed Students:', passed_count)

matrix = marks.reshape(2, 3)
print('2 × 3 Matrix:')
print(matrix)
```

```python
# activity C
import pandas as pd

data = {
    'student': ['Asha', 'Babu', 'Chayan', 'Dipa', 'Emon', 'Farah'],
    'study_hours': [2, 5, 1, 7, 3, 6],
    'sleep_hours': [8, 6, 9, 5, 7, 6],
    'attendance': [70, 88, 55, 95, 75, 91],
    'snacks': [3, 2, 5, 1, 4, 2],
    'passed': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

display(df)

print(df.info())

print(df.describe())

features = ['study_hours', 'sleep_hours', 'attendance', 'snacks']
label = 'passed'

print("Features:", features)
print("Label:", label)

print("Samples:", df.shape[0])
print("Columns:", df.shape[1])

high_attendance = df[df['attendance'] >= 80]
display(high_attendance)

passed_students = df[df['passed'] == 'Yes']
average_study_hours = passed_students['study_hours'].mean()

print("Average study hours of passed students:", average_study_hours)

sorted_df = df.sort_values('attendance', ascending=False)
display(sorted_df)

print("The dataset does not prove that snack purchases cause failure.")
print("It has only 6 students, and other factors like study hours, sleep, and attendance can affect the result.")

```

```python
import random
import numpy as np

def generate_scores(use_seed):
    if use_seed:
        random.seed(42)
        np.random.seed(42)

    return np.random.randint(40, 101, size=5)

print('Seeded run 1:', generate_scores(True))
print('Seeded run 2:', generate_scores(True))
print('Unseeded run 1:', generate_scores(False))
print('Unseeded run 2:', generate_scores(False))
```
